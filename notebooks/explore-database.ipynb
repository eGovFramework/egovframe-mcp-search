# eGovFrame MCP Server - 데이터베이스 탐색

하이브리드 데이터베이스에 저장된 데이터를 확인합니다.

**SQLite** (정형 데이터)
- `compatibility_software` - 호환성 소프트웨어
- `standard_terms` - 표준 용어 (6,400+)
- `standard_words` - 표준 단어 (1,000+)
- `standard_domains` - 표준 도메인 (100+)
- `common_components` - 공통컴포넌트 (154종)
- `component_categories` - 컴포넌트 카테고리
- `component_dependencies` - 컴포넌트 의존성

**LanceDB** (벡터 데이터)
- `developer_knowledge` - 개발자 지식베이스
- `egovframe_guide` - eGovFrame 공식 문서 (1,867 청크)

```python
import sqlite3
import lancedb
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 80)
pd.set_option('display.max_rows', 50)

KNOWLEDGE_BASE = '../knowledge-base'
SQLITE_PATH = f'{KNOWLEDGE_BASE}/egovframe.db'

print('=== SQLite ===')
conn = sqlite3.connect(SQLITE_PATH)
tables = pd.read_sql("SELECT name, type FROM sqlite_master WHERE type IN ('table','view') ORDER BY name", conn)
print(tables.to_string(index=False))

print('\n=== LanceDB ===')
ldb = lancedb.connect(KNOWLEDGE_BASE)
print('테이블:', ldb.table_names())
```

---
## Part 1. SQLite 정형 데이터

### 1-1. 테이블별 레코드 수

```python
main_tables = [
    'compatibility_software',
    'standard_terms',
    'standard_words',
    'standard_domains',
    'component_categories',
    'common_components',
    'component_dependencies',
]

for t in main_tables:
    count = conn.execute(f'SELECT COUNT(*) FROM {t}').fetchone()[0]
    print(f'  {t:35s} {count:>6,} rows')
```

### 1-2. 호환성 소프트웨어 (`compatibility_software`)

```python
df_compat = pd.read_sql('SELECT * FROM compatibility_software LIMIT 10', conn)
df_compat
```

```python
# 프레임워크 버전별 분포
print('=== 프레임워크 버전별 소프트웨어 수 ===')
df = pd.read_sql('''
    SELECT framework_version, COUNT(*) as cnt 
    FROM compatibility_software 
    GROUP BY framework_version 
    ORDER BY framework_version DESC
''', conn)
print(df.to_string(index=False))
```

```python
# 카테고리별 분포
print('=== 카테고리별 소프트웨어 수 ===')
df = pd.read_sql('''
    SELECT category, COUNT(*) as cnt 
    FROM compatibility_software 
    GROUP BY category 
    ORDER BY cnt DESC
''', conn)
print(df.to_string(index=False))
```

```python
# 호환성 레벨별 분포
print('=== 호환성 레벨별 분포 ===')
print('  Level 1: 적극 호환, Level 2: 제한적 호환')
df = pd.read_sql('''
    SELECT compatibility_level, COUNT(*) as cnt 
    FROM compatibility_software 
    GROUP BY compatibility_level
''', conn)
print(df.to_string(index=False))
```

### 1-3. 표준 용어 (`standard_terms`)

```python
df_terms = pd.read_sql('SELECT * FROM standard_terms LIMIT 10', conn)
df_terms
```

```python
# 도메인별 용어 수
print('=== 도메인별 표준 용어 수 (상위 20개) ===')
df = pd.read_sql('''
    SELECT domain_name, COUNT(*) as cnt 
    FROM standard_terms 
    WHERE domain_name IS NOT NULL AND domain_name != ''
    GROUP BY domain_name 
    ORDER BY cnt DESC 
    LIMIT 20
''', conn)
print(df.to_string(index=False))
```

```python
# 표준 용어 검색 예시
keyword = '게시판'
df = pd.read_sql(f'''
    SELECT korean_name, english_name, domain_name, description 
    FROM standard_terms 
    WHERE korean_name LIKE '%{keyword}%' 
    LIMIT 20
''', conn)
print(f'=== "{keyword}" 검색 결과 ({len(df)}건) ===')
df
```

### 1-4. 표준 단어 (`standard_words`)

```python
df_words = pd.read_sql('SELECT * FROM standard_words LIMIT 10', conn)
df_words
```

```python
# 형식 단어(is_format_word) 비율
df = pd.read_sql('''
    SELECT 
        CASE is_format_word WHEN 1 THEN '형식단어' ELSE '일반단어' END as type,
        COUNT(*) as cnt
    FROM standard_words
    GROUP BY is_format_word
''', conn)
print('=== 형식단어 vs 일반단어 ===')
print(df.to_string(index=False))
```

### 1-5. 표준 도메인 (`standard_domains`)

```python
df_domains = pd.read_sql('SELECT * FROM standard_domains LIMIT 10', conn)
df_domains
```

```python
# 데이터 타입별 도메인 분포
print('=== 데이터 타입별 도메인 수 ===')
df = pd.read_sql('''
    SELECT data_type, COUNT(*) as cnt 
    FROM standard_domains 
    GROUP BY data_type 
    ORDER BY cnt DESC
''', conn)
print(df.to_string(index=False))
```

### 1-6. 공통컴포넌트 (`common_components`)

```python
df_comp = pd.read_sql('''
    SELECT c.component_id, c.name, c.name_en, cat.name as category_name, 
           c.package_name, c.version, c.has_web
    FROM common_components c
    JOIN component_categories cat ON c.category_id = cat.category_id
    LIMIT 20
''', conn)
df_comp
```

```python
# 카테고리별 컴포넌트 수
print('=== 카테고리별 공통컴포넌트 수 ===')
df = pd.read_sql('''
    SELECT cat.name as category, COUNT(*) as cnt
    FROM common_components c
    JOIN component_categories cat ON c.category_id = cat.category_id
    GROUP BY cat.name
    ORDER BY cnt DESC
''', conn)
print(df.to_string(index=False))
```

```python
# 의존성이 가장 많은 컴포넌트
print('=== 의존성이 많은 컴포넌트 (상위 10개) ===')
df = pd.read_sql('''
    SELECT c.name, c.component_id, COUNT(d.depends_on_package) as dep_count
    FROM common_components c
    JOIN component_dependencies d ON c.component_id = d.component_id
    GROUP BY c.component_id
    ORDER BY dep_count DESC
    LIMIT 10
''', conn)
print(df.to_string(index=False))
```

```python
# 가장 많이 참조되는 컴포넌트 (다른 컴포넌트가 의존하는)
print('=== 가장 많이 참조되는 컴포넌트 패키지 (상위 10개) ===')
df = pd.read_sql('''
    SELECT depends_on_package, COUNT(*) as referenced_by
    FROM component_dependencies
    GROUP BY depends_on_package
    ORDER BY referenced_by DESC
    LIMIT 10
''', conn)
print(df.to_string(index=False))
```

### 1-7. FTS5 전체 텍스트 검색

```python
# 공통컴포넌트 FTS 검색 예시
keyword = '게시판'
df = pd.read_sql(f'''
    SELECT c.component_id, c.name, c.package_name, c.description
    FROM components_fts fts
    JOIN common_components c ON c.rowid = fts.rowid
    WHERE components_fts MATCH '{keyword}'
    LIMIT 10
''', conn)
print(f'=== FTS 검색: "{keyword}" ({len(df)}건) ===')
df
```

## Part 2. LanceDB 벡터 데이터

### 2-1. developer_knowledge 테이블

```python
tbl_knowledge = ldb.open_table('developer_knowledge')
df_knowledge = tbl_knowledge.to_pandas()

print(f'총 레코드 수: {len(df_knowledge)}')
print(f'컬럼: {list(df_knowledge.columns)}')
print(f'\n스키마:')
print(df_knowledge.dtypes)
```

```python
# vector 컬럼 제외하고 사용자 지식 데이터 확인
cols = [c for c in df_knowledge.columns if c != 'vector']
df_knowledge[cols]
```

```python
from sentence_transformers import SentenceTransformer
import pandas as pd

query = "호환성 신청은 어떻게 하나요?"

model = SentenceTransformer("intfloat/multilingual-e5-small")

# E5 계열은 query/passage prefix가 중요합니다.
query_vector = model.encode(
    [f"query: {query}"],
    normalize_embeddings=True
)[0].tolist()

results = tbl_knowledge.search(query_vector).limit(5).to_pandas()

cols = [c for c in results.columns if c != "vector"]

with pd.option_context("display.max_colwidth", None):
    display(results[cols])
```

```python
# type별 분포
if len(df_knowledge) > 0:
    print('=== type별 분포 ===')
    print(df_knowledge['type'].value_counts())
    print()
    print('=== importance별 분포 ===')
    print(df_knowledge['importance'].value_counts().sort_index())
    print()
    print('=== verified 분포 ===')
    print(df_knowledge['verified'].value_counts())
else:
    print('데이터가 없습니다.')
```

```python
# 태그 분석
if len(df_knowledge) > 0:
    all_tags = []
    for tags in df_knowledge['tags']:
        if hasattr(tags, 'tolist'):
            tags = tags.tolist()
        if isinstance(tags, list):
            all_tags.extend(tags)
    
    if all_tags:
        tag_counts = pd.Series(all_tags).value_counts()
        print(f'=== 태그 분포 (총 {len(tag_counts)}종) ===')
        print(tag_counts.head(20))
    else:
        print('태그가 없습니다.')
```

### 2-2. egovframe_guide 테이블

```python
tbl_guide = ldb.open_table('egovframe_guide')
df_guide = tbl_guide.to_pandas()

print(f'총 레코드 수: {len(df_guide)}')
print(f'컬럼: {list(df_guide.columns)}')
print(f'\n스키마:')
print(df_guide.dtypes)
```

```python
# vector 컬럼 제외하고 미리보기
cols = [c for c in df_guide.columns if c != 'vector']
df_guide[cols].head(20)
```

```python
# layer별 분포
if len(df_guide) > 0:
    print('=== layer별 분포 ===')
    print(df_guide['layer'].value_counts())
    print()
    print('=== category별 분포 (상위 20개) ===')
    print(df_guide['category'].value_counts().head(20))
else:
    print('데이터가 없습니다.')
```

```python
# 문서별 청크 수
if len(df_guide) > 0:
    doc_chunks = df_guide.groupby('docTitle').size().sort_values(ascending=False)
    print(f'=== 문서별 청크 수 (총 {len(doc_chunks)}개 문서) ===')
    print(doc_chunks.head(20))
```

```python
# 청크 내용 길이 분석
if len(df_guide) > 0:
    df_guide['content_len'] = df_guide['content'].str.len()
    print('=== 청크 내용 길이 통계 ===')
    print(df_guide['content_len'].describe())
```

### 2-3. 벡터 검색 테스트

```python
# egovframe_guide에서 첫 번째 레코드 벡터로 유사 검색
if len(df_guide) > 0:
    sample_vector = df_guide.iloc[0]['vector']
    sample_title = df_guide.iloc[0]['title']
    
    print(f'검색 기준: "{sample_title}"')
    print(f'벡터 차원: {len(sample_vector)}')
    print('---')
    
    results = tbl_guide.search(sample_vector).limit(5).to_pandas()
    for i, row in results.iterrows():
        print(f'{i+1}. [{row["layer"]}/{row["category"]}] {row["title"]}')
        print(f'   거리: {row["_distance"]:.4f}')
        print(f'   내용: {row["content"][:100]}...')
        print()
```
